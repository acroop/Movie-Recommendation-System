```python
import pandas as pd
import numpy as np
import ast

movies = pd.read_csv("../data/raw/tmdb_5000_movies.csv")
credits = pd.read_csv("../data/raw/tmdb_5000_credits.csv")
```

```python
movies.rename(columns={"id": "movie_id"}, inplace=True)

df = movies.merge(
    credits,
    on="movie_id",
    how="inner",
    suffixes=("", "_credits")
)

print("Merged dataset shape:", df.shape)
```

```python
df.head()
```

```python
df = df[
    [
        "movie_id",
        "title",
        "overview",
        "genres",
        "keywords",
        "cast",
        "crew"
    ]
].copy()

print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df.dropna(subset=["title"])

df["overview"] = df["overview"].fillna("")

df.reset_index(drop=True)
```

```python
def convert(text):
    try:
        data = ast.literal_eval(text)

        return [
            item["name"]
            for item in data
            if isinstance(item, dict) and "name" in item
        ]

    except (ValueError, SyntaxError, TypeError):
        return []
```

```python
df["genres"] = df["genres"].apply(convert)

df["keywords"] = df["keywords"].apply(convert)
```

```python
print(df[["title", "genres", "keywords"]].head())
```

```python
def extract_cast(text):
    try:
        data = ast.literal_eval(text)

        return [
            actor["name"]
            for actor in data[:3]
            if isinstance(actor, dict) and "name" in actor
        ]

    except (ValueError, SyntaxError, TypeError):
        return []
```

```python
df["cast"] = df["cast"].apply(extract_cast)
```

```python
def extract_director(text):
    try:
        data = ast.literal_eval(text)

        for person in data:
            if (
                isinstance(person, dict)
                and person.get("job") == "Director"
            ):
                return [person["name"]]

        return []

    except (ValueError, SyntaxError, TypeError):
        return []
```

```python
print(df.columns)
```

```python
df["crew"] = df["crew"].apply(extract_director)

df.rename(columns={"crew": "director"}, inplace=True)
```

```python
df.columns
```

```python
def normalize(items):
    return [
        str(item).lower().replace(" ", "")
        for item in items
    ]
```

```python
df["genres"] = df["genres"].apply(normalize)
df["keywords"] = df["keywords"].apply(normalize)
df["cast"] = df["cast"].apply(normalize)
df["director"] = df["director"].apply(normalize)
```

```python
df["overview"] = df["overview"].apply(
    lambda x: x.lower().split()
)
```

```python
df["tags"] = (
    df["overview"]
    + df["genres"]
    + df["keywords"]
    + df["cast"]
    + df["director"]
)
```

```python
df["tags"] = df["tags"].apply(
    lambda x: " ".join(x)
)
```

```python
print(df[["title", "tags"]].head())
```

```python
final_df = df[
    ["movie_id", "title", "tags"]
].copy()
```

```python
final_df.drop_duplicates(
    subset=["movie_id"],
    inplace=True
)

final_df.reset_index(drop=True, inplace=True)
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

final_df.to_csv(
    "../data/processed/movies_cleaned.csv",
    index=False
)

print("Processed dataset saved successfully!")
print(final_df.shape)
```
